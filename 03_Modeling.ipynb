## 0. 환경 설정 및 데이터 불러오기

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
model_df = pd.read_csv("/Users/CHOIEUNSEO/Desktop/AI·AX 전문가 과정/프로젝트/data/model_df.csv")
```

```python
# 1. 독립변수(X)와 종속변수(Y) 분리
# model_df에 '거래량'과 파생변수들이 정제되어 있는 상태여야 합니다.
Y = model_df['거래량']
X = model_df.drop(columns=['거래량'])

# 회귀식의 절편(b)을 계산하기 위해 필수적으로 상수항을 추가합니다.
X_with_constant = sm.add_constant(X)

# 2. 다중선형회귀모델(OLS) 학습 및 결과 요약
model = sm.OLS(Y, X_with_constant).fit()
print(model.summary())
```
